# Qibocal executions, displayed implicitly

Every phase of a `JupyterExecutor` returns a `JupyterCompleted`, which renders itself when it is the last expression of a cell.

Point `QIBOLAB_PLATFORMS` to a folder containing your platform (e.g. the `tests/platforms` folder of the Qibocal repository for the `mock` one).

```python
import os
import shutil
from pathlib import Path

from qibocal import protocols

from qibocal_widgets.jupyter import JupyterExecutor

PLATFORM = os.environ.get("QIBOCAL_WIDGETS_PLATFORM", "mock")
path = Path("var/report")
if path.exists():
    shutil.rmtree(path)
executor = JupyterExecutor.create(path=path, platform=PLATFORM, update=False)
```

```python
bound = protocols.ramsey_signal(
    delay_between_pulses_start=10,
    delay_between_pulses_end=1000,
    delay_between_pulses_step=50,
)
with executor:
    acquired = executor.acquire(bound, targets=[0, 1])
acquired
```

```python
fitted = executor.fit(acquired)
fitted.results
```

```python
reported = executor.report(fitted)
reported
```

## Explicit helpers

The same content is available through methods and functions, for custom layouts.

```python
import ipywidgets as widgets

widgets.HBox([reported.show_data(), reported.show_reports()])
```
